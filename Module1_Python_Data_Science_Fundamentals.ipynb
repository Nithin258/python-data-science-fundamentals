## Day 2 – Control flow & Functions

```python
# Loops and conditions
for n in range(1, 11):
    if n % 2 == 0:
        print(n, 'is even')
    else:
        print(n, 'is odd')

total, i = 0, 1
while i <= 5:
    total += i
    i += 1
print('Sum 1..5 =', total)
```

```python
# Functions
def average(numbers):
    """Return the mean of a list of numbers."""
    return sum(numbers) / len(numbers) if numbers else 0

def grade(score):
    if score >= 90: return 'A'
    if score >= 75: return 'B'
    if score >= 60: return 'C'
    return 'D'

marks = [88, 92, 79, 65, 95]
print('Average:', average(marks))
print([grade(m) for m in marks])
```

## Day 3 – Lists, Dictionaries & File Handling

```python
# Lists
fruits = ['apple', 'banana', 'cherry']
fruits.append('mango'); fruits.remove('banana')
print(fruits, '| sorted:', sorted(fruits), '| first two:', fruits[:2])
squares = [x**2 for x in range(1, 8)]
print(squares)

# Dictionaries
student = {'name': 'Asha', 'marks': {'Math': 90, 'Physics': 82, 'CS': 95}}
student['marks']['English'] = 78
for subject, m in student['marks'].items():
    print(f'{subject:8} {m}')
print('Best subject:', max(student['marks'], key=student['marks'].get))
```

```python
# File handling
import csv
rows = [['name','score'], ['Asha',90], ['Ravi',72], ['Meena',85], ['John',60]]
with open('scores.csv', 'w', newline='') as f:
    csv.writer(f).writerows(rows)

with open('scores.csv') as f:
    data = list(csv.DictReader(f))
print(data)

with open('notes.txt', 'w') as f:
    f.write('Line 1\nLine 2\n')
with open('notes.txt') as f:
    print(f.read())
```

## Day 4 – Intro to Data Science & Real-World Applications
Data Science workflow: **Collect → Clean → Explore → Visualize → Model → Communicate**. Real-world uses: recommendation systems, fraud detection, healthcare prediction, weather forecasting, sales analytics.

```python
import statistics as st
scores = [int(r['score']) for r in data]
print('Mean:', st.mean(scores), '| Median:', st.median(scores), '| Std dev:', round(st.stdev(scores), 2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('scores.csv')
df['passed'] = df['score'] >= 70
print(df)
print(df.describe())

df.plot(kind='bar', x='name', y='score', legend=False, color='teal')
plt.title('Student Scores'); plt.ylabel('Score'); plt.tight_layout(); plt.show()
```
